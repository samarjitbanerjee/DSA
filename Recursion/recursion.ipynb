#### Simple Recursion Function for understanding

```python
def recursion_example(n):
    """A simple recursive function that prints numbers from n to 0."""
    ### base case to stop recursion
    ### if n is equal to 0, the function will return and stop calling itself
    if n == 0:
        print(n)
        return
    print(n)
    recursion_example(n - 1)

recursion_example(5)

### inside stack memory first main function(recursion_example) is called
### then the function calls itself with n-1 until n is equal to 0
### each call is added to the stack memory
### when n is equal to 0, the function returns and the stack unwinds
### the output will be 5, 4, 3, 2, 1
```

#### Find nth Fibonacci Number

```python
def fibonacci(n: int) -> int:
    """A recursive function to find the nth Fibonacci number."""
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    
    return fibonacci(n - 1) + fibonacci(n - 2)
    
fibonacci(6)
```
